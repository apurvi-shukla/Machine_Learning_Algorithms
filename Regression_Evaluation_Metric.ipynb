```python
import pandas as pd
```

```python
# Creating Dataset

df = pd.DataFrame({
    "StudyHours": [1, 2, 3, 4, 5, 6, 7, 8],
    "Marks":      [30, 35, 50, 55, 65, 70, 80, 85]
})

X = df[["StudyHours"]]
y = df["Marks"]

print(df)
```

```python
# Importing Linear Regression
# y = b0 + b1 * x

from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X, y) # Model learn from the given data

print("Slope:", model.coef_[0]) # Slope: change in predicted marks per study hour
print("Intercept:", model.intercept_)  # Intercept: predicted marks when study hours = 0
```

```python
new_student = pd.DataFrame({
    "StudyHours": [9]
})

predicted_marks = model.predict(new_student) # Marks to be Predicted

print("Predicted Marks:", predicted_marks[0])
```

```python
predictions = model.predict(X) # To Predict

results = pd.DataFrame({
    "StudyHours": df["StudyHours"],
    "ActualMarks": y,
    "PredictedMarks": predictions
})

results["Error"] = (
    results["ActualMarks"] - results["PredictedMarks"]
)

results["AbsoluteError"] = results["Error"].abs() # Removing Negative sign from Error

print(results.round(2)) # Round off the decimal value to 2
```

```python
#  Calculating Mean Absolute Error

from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(y, predictions)

print("MAE:", round(mae, 2))
```

```python
#  Calculating Mean Squared Error

from sklearn.metrics import mean_squared_error

mse = mean_squared_error(y, predictions)

print("MSE:", round(mse, 2))
```

```python
# Calculating Root Mean Squared Error

import numpy as np # Importing Numpy for Mathematical calculation [square root]

rmse = np.sqrt(mse)

print("RMSE:", round(rmse, 2))
```

```python
#  Calculating R-Sqaured Error

from sklearn.metrics import r2_score

r2 = r2_score(y, predictions)

print("R² Score:", round(r2, 4))
```
