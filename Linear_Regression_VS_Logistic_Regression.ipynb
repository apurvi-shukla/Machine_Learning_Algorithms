```python
# Linear Rgression = Regression [Numeric Values]
# Logistic Regression = Classification [Classes / Binary Output]
```

```python
import pandas as pd
```

```python
# Creating Dataset

# Linear Regression

df = pd.DataFrame({
    "StudyHours": [1, 2, 3, 4, 5, 6, 7, 8],
    "Marks":      [30, 35, 50, 55, 65, 70, 80, 85]
})

print(df)
```

```python
# Dividing Dataset into Features and Target

X = df[["StudyHours"]] # Input
y = df["Marks"] # Output

print("Features (X):")
print(X)

print("\nTarget (y):")
print(y)
```

```python
from sklearn.linear_model import LinearRegression # Importing Linear Regression Algorithm

model = LinearRegression()
model.fit(X, y) # Making Dataset Learn from the given Inputs

new_student = pd.DataFrame({
    "StudyHours": [9]
})

prediction = model.predict(new_student) # used for the prediction of Marks for 9 Hours

print("Predicted marks:", round(prediction[0], 2)) # printing Final Output
```

```python
# Creating Dataset

# Logistic Regression

df = pd.DataFrame({
    "Marks":      [25, 35, 45, 60, 75, 30, 55, 40],
    "Attendance": [50, 60, 70, 75, 90, 40, 80, 55],
    "Result": [
        "Fail", "Fail", "Pass", "Pass",
        "Pass", "Fail", "Pass", "Fail"
    ]
})

print(df)
```

```python
# Dividing Dataset into Features and Target

X = df[["Marks", "Attendance"]] # Input
y = df["Result"] # Output

print("Features:")
print(X)

print("\nTarget:")
print(y)
```

```python
from sklearn.linear_model import LogisticRegression # Importing Logistic Regression Algorithm

model = LogisticRegression()
model.fit(X, y) # Making Dataset Learn from the given Inputs

new_student = pd.DataFrame({
    "Marks": [50],
    "Attendance": [70]
})

prediction = model.predict(new_student) # used for the prediction of Result with given Marks=50 and Attendance=70

print("Predicted Result:", prediction[0]) # printing Final Output
```
