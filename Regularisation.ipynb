```python
import pandas as pd
```

```python
# Creating a DataFrame

df = pd.DataFrame({
    "StudyHours": [1, 2, 3, 4, 5, 6, 7, 8],
    "Marks":      [30, 35, 50, 55, 65, 70, 80, 85]
})

X = df[["StudyHours"]] # Features
y = df["Marks"] # Target

print(df)
```

```python
# Importing Lasso Regression : L1 Regularisation

from sklearn.linear_model import Lasso

lasso_model = Lasso(alpha=1.0) # Creating a Model

lasso_model.fit(X, y) # Training Model

predictions = lasso_model.predict(X) # Predicting the Marks

print(predictions) # Print the Predicted Marks
```

```python
# Importing Ridge Regression : L2 Regularisation

from sklearn.linear_model import Ridge

ridge_model = Ridge(alpha=1.0) # Creating a Model

ridge_model.fit(X, y) # Training Model

predictions = ridge_model.predict(X) # Predicting the Marks

print(predictions) # Print the Predicted Marks
```
