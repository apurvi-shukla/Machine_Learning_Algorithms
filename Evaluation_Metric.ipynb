```python
# Creating Confusion Matrix

from sklearn.metrics import confusion_matrix

y_actu = [1, 1, 0, 0, 1, 0, 1, 0] # Actual Values
y_pred = [1, 0, 0, 0, 1, 1, 1, 0] # Predicted Values

cm = confusion_matrix(y_actu, y_pred)

print("Confusion Matrix:")
print(cm)
```

```python
# Measuring Accuracy

from sklearn.metrics import accuracy_score

y_true = [1, 1, 0, 0, 1, 0, 1, 0] # Actual Values
y_pred = [1, 0, 0, 0, 1, 1, 1, 0] # Predicted Values

accuracy = accuracy_score(y_true, y_pred)

print("Accuracy:", accuracy)
print("Accuracy (%):", accuracy * 100) # Converting accuracy into %
```

```python
# Measuring Precission

from sklearn.metrics import precision_score

y_true = [1, 1, 0, 0, 1, 0, 1, 0] # Actual Values
y_pred = [1, 0, 0, 0, 1, 1, 1, 0] # Predicted Values

precision = precision_score(y_true, y_pred)

print("Precision:", precision)
print("Precision (%):", precision * 100) # Converting precission into %
```

```python
# Measuring Recall

from sklearn.metrics import recall_score

y_true = [1, 1, 0, 0, 1, 0, 1, 0] # Actual Values
y_pred = [1, 0, 0, 0, 1, 1, 1, 0] # Predicted Values

recall = recall_score(y_true, y_pred)

print("Recall:", recall)
print("Recall (%):", recall * 100) # Converting recall into %
```

```python
# Measuring F1-score

from sklearn.metrics import f1_score

y_true = [1, 1, 0, 0, 1, 0, 1, 0] # Actual Values
y_pred = [1, 0, 0, 0, 1, 1, 1, 0] # Predicted Values

f1 = f1_score(y_true, y_pred)

print("F1 Score:", f1)
print("F1 Score (%):", f1 * 100) # Converting F1-score into %
```
